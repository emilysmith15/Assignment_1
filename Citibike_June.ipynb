```python
# Now I'm going to write some code that will make the datasets match.
```

```python
## First, I am going to change the duration column. This dataset already has a duration column, but it measures duration differently than
## the other data, and I think it would just be easier to drop it and calculate the duration using the "stoptime" and "starttime" columns. 

# Dropping the original duration column.
data_2019_duration = data_2019_concat.drop(columns=['tripduration'])

# Calculating duration.
data_2019_duration["duration"] = data_2019_duration["stoptime"] - data_2019_duration["starttime"]
```

```python
# Now, I want to rename the columns so they match the data from previous years. 

# Columns:
data_2019_columns = data_2019_concat.rename(columns = {
    "starttime":"started_at", 
    "stoptime":"ended_at", 
    "start station id":"start_station_id", 
    "start station name":"start_station_name",
    "start station latitude":"start_lat", 
    "start station longitude":"start_lng", 
    "end station id":"end_station_id", 
    "end station name":
    "end_station_name", 
    "end station latitude":"end_lat", 
    "end station longitude":"end_lng", 
    "usertype":"member_casual", 
    "birth year":
    "birth_year"})

# Now I'm going to fix the member_casual column so that the options within the cells match the options in the previous data. 

data_2019_columns["member_casual"] = data_2019_columns["member_casual"].replace({
    "Subscriber":"member",
    "Customer":"casual"})

# Now I'm going to put them in the same order as the ones above, excluding columns that aren't included in the previous data.

data_2019 = data_2019_columns.reindex(columns=["started_at", "ended_at", "duration", "start_station_name", "start_station_id", 
                                               "end_station_name", "end_station_id", "start_lat", "start_lng", "end_lat", "end_lng",
                                               "member_casual", "rideable_type"])

#Ok, now I think I have pretty comparable data sets. 
```
